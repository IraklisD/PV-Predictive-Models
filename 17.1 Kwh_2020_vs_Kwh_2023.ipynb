```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("D:\Σπουδές\Μαθήματα-Σημειώσεις Data Science\Dissertation\Data\cleaned_data_new_csv.csv", index_col= "datetime", parse_dates= True)
data.head()
```

```python
plt.figure(figsize= (16, 9))
sns.set_style("darkgrid")
sns.lineplot(x = data.index, y = data['temp'])
plt.show()
```

```python
old = data[data.index < "2021-01-01 00:00:00"]
new = data[(data.index >= "2023-01-01 00:00:00") & (data.index < "2024-01-01 00:00:00")]
```

```python
old['temp'].mean(), new['temp'].mean()
```

```python
old["solarradiation"].mean(), new["solarradiation"].mean()
```

```python
old['Kwh'][old['Kwh'] != 0].mean(), new['Kwh'][new['Kwh'] != 0].mean()
```

```python
old_mean_hours = old.groupby(old.index.hour)['Kwh'].mean()
new_mean_hours = new.groupby(new.index.hour)['Kwh'].mean()

means_hours_df = pd.DataFrame({
    "Old Means": old_mean_hours,
    "New Means": new_mean_hours
})
```
