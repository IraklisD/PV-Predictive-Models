# Weather

```python
weeather = pd.read_csv("Kozani, Weather 2020-01-01 to 2024-06-26.csv")
weeather
```

```python
energy = pd.read_csv("final_data.csv")
energy
```

```python
kairos = pd.read_csv("Kozani, Greece 2024-06-27 to 2024-07-04.csv")
kairos
```

```python
weather = pd.concat([weeather, kairos], axis= 0, ignore_index= True)
weather
```

```python
weather.iloc[17544:]
```

```python
energy['Date'] = pd.to_datetime(energy['Date'], errors='coerce', dayfirst=True)
weather['datetime'] = pd.to_datetime(weather['datetime'], errors='coerce')
```

```python
energy["Date"], weather.iloc[17544:]
```

```python
merged_df = pd.merge(energy, weather.iloc[17544:], left_on='Date', right_on='datetime', how='outer')
merged_df
```

```python
merged_df[merged_df["Date"] == "2022-03-02 20:00:00"]
```

```python
final_df = merged_df.sort_values(by = "datetime")
final_df
```

```python
final_df["Date"][pd.isna(final_df["Date"])]
```

```python
final_df["Date"][pd.isna(final_df["Date"])] = final_df["datetime"][pd.isna(final_df["Date"])]
final_df
```

```python
final_df[pd.isna(final_df["Kwh"])]
```

```python
final_df["Kwh"] = final_df["Kwh"].fillna(0)
final_df
```

```python
final_df.isna().sum()
```

```python
final_df.to_csv("merged_df.csv", index= False)

final_df
```
