```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
models = ["SARIMA", "Random Forest", "SVR", "Simple LSTM", "LSTM-Attention", "CNN-LSTM\nAttention", "TDNN", "TDNN-Attention"]
r_squared_scores = [0.9369, 0.8586, 0.8519, 0.8926, 0.9377, 0.9528, 0.9506, 0.9536]
mse_scores = [0.00401, 0.0087, 0.0097, 0.0066, 0.0038, 0.0031, 0.0031, 0.003]
time = [2.274, 0.0556, 0.0697, 0.0788, 2.054, 9.562, 0.471, 6.51]
```

```python
results_df = pd.DataFrame({"Model": models, "R-Squared": r_squared_scores, "MSE" : mse_scores, "Time" : time})
results_df
```

```python
custom_palette = [
    "#8dd3c7", "#ffffb3", "#bebada", "#fb8072", 
    "#80b1d3", "#fdb462", "#b3de69", "#fccde5"
]
```

```python
plt.figure(figsize=(16, 9))
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")

sns.barplot(
    x="Model", 
    y="R-Squared", 
    data=results_df, 
    palette=custom_palette, 
    edgecolor='black',
    width= 0.7
)
plt.xlabel("Models")
plt.ylabel(r"$R^2$ Score")

plt.rcParams.update({
    'font.size': 10,   
    'axes.labelsize': 10,  
    'axes.titlesize': 10,  
    'xtick.labelsize': 10,  
    'ytick.labelsize': 10, 
    'legend.fontsize': 10,    
    'figure.titlesize': 10,    
    'axes.titleweight': 'bold',  
    'axes.labelweight': 'bold', 
    'font.family': 'serif',     
    'figure.figsize': (16, 9),    
})

for index, row in results_df.iterrows():
    plt.text(
        x=index, 
        y=row['R-Squared'] + 0.01,  
        s=f"{row['R-Squared']:.4f}",  
        ha='center', 
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )

plt.xticks(rotation=45)
plt.ylim(0.5, 1.0)  

plt.show()
```

```python
plt.figure(figsize=(16, 9))
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")
sns.barplot(x = "Model", 
            y = "MSE", 
            data= results_df, 
            palette=custom_palette,
            edgecolor='black',
            width= 0.7
            )
plt.xlabel("Models")
plt.ylabel("Mean Squared Error")

for index, row in results_df.iterrows():
    plt.text(
        x=index, 
        y=row['MSE'] + 0.000005,  
        s=f"{row['MSE']:.4f}",  
        ha='center', 
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )
plt.xticks(rotation = 45)

plt.show()
```

### This means that our model excludes 3 watts on average from the actual value

```python
plt.figure(figsize=(16, 9))
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")
sns.barplot(x = "Model", 
            y = "Time", 
            data= results_df, 
            palette=custom_palette,
            edgecolor='black',
            width= 0.6
            )
plt.xlabel("Models", fontsize= 20)
plt.ylabel("Training Time (Mins)", fontsize = 24)

for index, row in results_df.iterrows():
    plt.text(
        x=index, 
        y=row["Time"] + 0.01,        
        s=f"{row['Time']:.2f}",  
        ha='center', 
        va='bottom',
        fontsize=20,
        fontweight='bold'
    )
plt.xticks(rotation = 90, fontsize = 24)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Setup
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")

# Create vertically stacked subplots
fig, (ax1, ax2) = plt.subplots(nrows=2, sharex=True, figsize=(16, 12))

# Plot R-Squared
sns.barplot(
    x="Model", 
    y="R-Squared", 
    data=results_df, 
    palette=custom_palette, 
    edgecolor='black',
    width=0.6,
    ax=ax1
)
ax1.set_ylabel(r"$R^2$ Score", fontsize=24, fontweight='bold')
"""ax1.set_title("Model Comparison: $R^2$ Score", fontsize=16, fontweight='bold')
"""
# Annotate R-Squared values
for index, row in results_df.iterrows():
    ax1.text(
        x=index,
        y=row['R-Squared'] + 0.01,
        s=f"{row['R-Squared']:.3f}",
        ha='center',
        va='bottom',
        fontsize=20,
        fontweight='bold'
    )
ax1.set_ylim(0.5, 1.0)

# Plot MSE
sns.barplot(
    x="Model", 
    y="MSE", 
    data=results_df, 
    palette=custom_palette, 
    edgecolor='black',
    width=0.6,
    ax=ax2
)
ax2.set_ylabel("Mean Squared Error", fontsize=24, fontweight='bold')
"""ax2.set_title("Model Comparison: Mean Squared Error", fontsize=16, fontweight='bold')
"""
# Annotate MSE values
for index, row in results_df.iterrows():
    ax2.text(
        x=index,
        y=row['MSE'] + 0.000005,
        s=f"{row['MSE']:.3f}",
        ha='center',
        va='bottom',
        fontsize=20,
        fontweight='bold'
    )

# Shared X axis label and formatting
ax2.set_xlabel("Models", fontsize=20, fontweight='bold')
ax2.set_xticklabels(results_df["Model"], rotation=90, ha='right', fontsize = 24)

# Layout adjustment
plt.tight_layout()
plt.show()
```
