```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
models = ["SARIMA", "Random Forest", "SVR", "Simple LSTM", "LSTM-Attention", "CNN-LSTM-Attention", "Time Delay", "Time Delay-Attention"]
r_squared_scores = [0.9369, 0.8586, 0.8519, 0.8926, 0.9377, 0.9528, 0.9506, 0.9536]
mse_scores = [0.00401, 0.0087, 0.0097, 0.0066, 0.0038, 0.0031, 0.0031, 0.003]
#time = [, , , , , 7.921, 7.521, ]
```

```python
results_df = pd.DataFrame({"Model": models, "R-Squared": r_squared_scores, "MSE" : mse_scores})
results_df
```

```python
plt.figure(figsize=(16, 9))
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")
sns.barplot(x = "Model", y = "R-Squared", data= results_df, palette= 'cividis', hatch = '/')
plt.xlabel("Models")
plt.ylabel(r"$R^2$ Score")
#plt.title(r"$R^2$ Score Per Model")

for index, row in results_df.iterrows():
    plt.text(
        x=index, 
        y=row['R-Squared'] + 0.01,  
        s=f"{row['R-Squared']:.4f}",  
        ha='center', 
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )
plt.xticks(rotation = 45)

plt.show()
```

```python
plt.figure(figsize=(16, 9))
sns.set_context("notebook", font_scale=1.5)
sns.set_style("whitegrid")
sns.barplot(x = "Model", y = "MSE", data= results_df, palette= 'cividis', hatch = '/')
plt.xlabel("Models")
plt.ylabel("Mean Squared Error")
#plt.title("Mean Squared Error Per Model")

for index, row in results_df.iterrows():
    plt.text(
        x=index, 
        y=row['MSE'] + 0.000005,  
        s=f"{row['MSE']:.4f}",  
        ha='center', 
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )
plt.xticks(rotation = 45)

plt.show()
```
